## Explication data et dictionnaire

<b>Data</b> -> la quantité d'ingrédient en grammes pour 1kg de produit

<b>Poids_unite</b> -> combien de grammes pour 1 produit

<b>df_unite</b> -> quantité d'ingrédient en kg pour chaque produit

<b>prix_ingredients</b> -> prix au kg de chaque ingrédient

---

L'eau qui fait partie de la recette de chaque produit n'a pas été pris en compte.

```python
import pandas as pd
```

```python
data = {
    "BAGUETTE": {"farine": 640, "sel": 18, "levure": 2},
    "TRADITIONAL_BAGUETTE": {"farine": 640, "sel": 18, "levure": 2},
    "PAIN_AU_CHOCOLAT": {"farine": 450, "sel": 10, "levure": 20, "sucre": 60, "beurre": 220, "lait": 20, "barres_chocolat": 20},
    "CROISSANT": {"farine": 470, "sel": 10, "levure": 20, "sucre": 50, "beurre": 220, "lait": 20},
    "BANETTE": {"farine": 630, "sel": 18, "levure": 2},
    "SPECIAL_BREAD": {"farine": 550, "sel": 18, "levure": 2, "graines": 70},
    "CAMPAGNE": {"farine": 500, "farine_de_seigle": 150, "sel": 20, "levure": 10},
    "CEREAL_BAGUETTE": {"farine": 600, "sel": 18, "levure": 2, "graines": 60},
    "VIK_BREAD": {"farine": 400, "farine_de_seigle": 200, "sel": 18, "levure": 5, "son_de_ble": 7},
    "ECLAIR": {"farine": 300, "beurre": 200, "sucre": 40, "sel": 10, "creme": 300, "oeufs": 150},
    "COOKIE": {"farine": 550, "sel": 5, "sucre": 300, "beurre": 300, "oeufs": 100, "chocolat": 200},
    "BOULE_400G": {"farine": 650, "sel": 18, "levure": 2},
    "TARTELETTE": {"farine": 500, "sucre": 200, "beurre": 250, "oeufs": 150, "creme": 200, "sel": 5},
}

df = pd.DataFrame.from_dict(data, orient="index").fillna(0)

poids_unite = {
    "BAGUETTE": 0.25,
    "TRADITIONAL_BAGUETTE": 0.25,
    "PAIN_AU_CHOCOLAT": 0.08,
    "CROISSANT": 0.07,
    "BANETTE": 0.25,
    "SPECIAL_BREAD": 0.50,
    "CAMPAGNE": 0.50,
    "CEREAL_BAGUETTE": 0.25,
    "VIK_BREAD": 0.50,
    "ECLAIR": 0.10,
    "COOKIE": 0.05,
    "BOULE_400G": 0.40,
    "TARTELETTE": 0.12,
}

df_unite = df.mul(pd.Series(poids_unite), axis=0) / 1000

prix_ingredients = {
    "farine": 1.2,
    "farine_de_seigle": 1.4,
    "sel": 0.20,
    "levure": 6.0,
    "sucre": 1.0,
    "beurre": 8.0,
    "lait": 0.9,
    "barres_chocolat": 6.5,
    "ameliorants": 4.0,
    "graines_speciales": 3.5,
    "son_de_ble": 1.0,
    "chocolat": 6.0,
    "oeufs": 2.5,
    "creme": 3.0,
}

df_prix = pd.DataFrame.from_dict(prix_ingredients, orient="index", columns=["prix_kg"])
```
